```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
student = pd.read_csv("./sample_data/sns_data.csv")
student.head()
```

```python
sns.lineplot(data=student, x = 'week',y='attendance_rate',errorbar=None)
```

```python
sns.lineplot(data=student, x = 'week',y='attendance_rate',errorbar=None, hue="class_level")
```

```python
#figure lavel
sns.relplot(kind='line', data=student, x='week',y='attendance_rate',errorbar=None,hue='gender')
```

```python
sns.relplot(kind='line', data=student, x='week',y='attendance_rate',errorbar=None,hue='gender',style='gender')
```

#scatter plot using seaborn


```python
sns.relplot(kind="scatter",data = student,x='study_hours', y='test_score', hue="gender",style="subject",size="Tshirt_size")
```

```python
tips_data = sns.load_dataset("tips")
tips_data.head()
```

```python
sns.relplot(kind="scatter", data=tips_data, x="total_bill", y="tip", hue="sex", style="time", size="size")
```

```python
df=pd.read_csv("./sample_data/student_dataset_complete.csv")
df
```

#facet plot

```python
sns.scatterplot(data=df, x='study_hours', y="test_score", hue="gender")
```

```python
sns.relplot(kind="scatter",data=df, x='study_hours', y="test_score", hue="gender")
```

```python
sns.relplot(kind="scatter",data=df, x='study_hours', y="test_score", col="gender")
```

```python
sns.relplot(kind="scatter",data=df, x='study_hours', y="test_score", col="gender", row="hostel")
```

```python
sns.relplot(kind="scatter",data=df, x='study_hours', y="test_score", col="week",col_wrap=2)
```

**histrogram using seaborn**



```python
sns.histplot(data=df, x="attendance_rate",bins=10, hue="gender",element="step")
```

```python
#figure level
sns.displot(data=df, x="attendance_rate",col="gender" )
```

```python
sns.displot(data=df, x="study_hours",col="gender" )
```
